# 01 - Credit Risk Exploratory Data Analysis
## BFSI Risk Intelligence Platform

This notebook explores the credit risk prediction dataset, including:
- Data loading and basic statistics
- Feature distributions
- Missing values analysis
- Target variable distribution
- Key insights for modeling

```python
# Import required libraries
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add src to path
sys.path.insert(0, os.path.join(os.getcwd(), '..'))

from src.config import CREDIT_RISK_RAW_FILE
from src.utils.common import load_data, get_data_info

# Set visualization style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## 1. Load and Explore Data

```python
# Load credit risk data
df = load_data(CREDIT_RISK_RAW_FILE)
print(f"Dataset shape: {df.shape}")
print(f"\nFirst few rows:")
df.head()
```

```python
# Get data info
get_data_info(df)
```

## 2. Analyze Target Variable

```python
# Check target distribution
print("Risk Distribution:")
print(df['risk'].value_counts())
print(f"\nPercentages:")
print(df['risk'].value_counts(normalize=True) * 100)

# Visualize
plt.figure(figsize=(8, 5))
df['risk'].value_counts().plot(kind='bar', color=['green', 'orange', 'red'])
plt.title('Target Variable Distribution - Risk Categories')
plt.xlabel('Risk Category')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

## 3. Feature Analysis

```python
# Numeric features
numeric_features = df.select_dtypes(include=[np.number]).columns.tolist()
print(f"Numeric features: {numeric_features}")

# Categorical features
categorical_features = df.select_dtypes(include=['object']).columns.tolist()
print(f"Categorical features: {categorical_features}")

# Describe numeric features
print("\nNumeric Features Statistics:")
df[numeric_features].describe()
```

```python
# Visualize numeric feature distributions
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
for idx, col in enumerate(numeric_features[:4]):
    ax = axes[idx//2, idx%2]
    df[col].hist(bins=30, ax=ax, color='skyblue', edgecolor='black')
    ax.set_title(f'Distribution of {col}')
    ax.set_xlabel(col)
    ax.set_ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
# Analyze categorical features
for col in categorical_features:
    if col != 'risk':
        print(f"\n{col}:")
        print(df[col].value_counts())
        print("-" * 50)
```

## 4. Missing Values Analysis

```python
# Check missing values
missing = df.isnull().sum()
if missing.sum() > 0:
    print("Missing values:")
    print(missing[missing > 0])
    print(f"\nPercentage of missing values:")
    print((missing[missing > 0] / len(df) * 100).round(2))
else:
    print("No missing values found!")
```
