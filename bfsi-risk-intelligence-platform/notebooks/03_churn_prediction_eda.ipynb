```python
# Import required libraries
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add src to path
sys.path.insert(0, os.path.join(os.getcwd(), '..'))

from src.config import CHURN_PREDICTION_RAW_FILE
from src.utils.common import load_data, get_data_info

# Set visualization style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## 1. Load and Explore Data

```python
# Load churn data
df = load_data(CHURN_PREDICTION_RAW_FILE)
print(f"Dataset shape: {df.shape}")
print(f"\nFirst few rows:")
df.head()
```

```python
# Get data info
get_data_info(df)
```

## 2. Analyze Churn Distribution

```python
# Check churn distribution
print("Churn Distribution:")
print(df['exited'].value_counts())
print(f"\nPercentages:")
print(df['exited'].value_counts(normalize=True) * 100)
print(f"\nChurn Rate: {df['exited'].mean()*100:.2f}%")

# Visualize
plt.figure(figsize=(8, 5))
colors = ['green', 'red']
df['exited'].value_counts().plot(kind='bar', color=colors)
plt.title('Customer Churn Distribution')
plt.xlabel('Exited (Churned)')
plt.ylabel('Count')
plt.xticks([0, 1], ['Retained', 'Churned'], rotation=0)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

## 3. Feature Analysis by Churn Status

```python
# Compare features by churn status
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
numeric_cols.remove('exited')

comparison = pd.DataFrame({
    'Retained_Mean': df[df['exited']==0][numeric_cols].mean(),
    'Churned_Mean': df[df['exited']==1][numeric_cols].mean(),
})
comparison['Difference'] = comparison['Churned_Mean'] - comparison['Retained_Mean']
print("Feature Comparison - Churned vs Retained Customers:")
print(comparison.round(2))
```
