# 04 - Loan Collection Risk Exploratory Data Analysis
## BFSI Risk Intelligence Platform

This notebook explores the collection risk dataset, including:
- Data loading and basic statistics
- Feature distributions
- Collection risk distribution
- EMI and income relationships
- Key predictors for collection priority

```python
# Import required libraries
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add src to path
sys.path.insert(0, os.path.join(os.getcwd(), '..'))

from src.config import COLLECTION_RISK_RAW_FILE
from src.utils.common import load_data, get_data_info

# Set visualization style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## 1. Load and Explore Data

```python
# Load collection risk data
df = load_data(COLLECTION_RISK_RAW_FILE)
print(f"Dataset shape: {df.shape}")
print(f"\nFirst few rows:")
df.head()
```

```python
# Get data info
get_data_info(df)
```

## 2. Analyze Collection Risk Distribution

```python
# Check collection risk distribution
print("Collection Risk Distribution:")
print(df['collection_risk'].value_counts())
print(f"\nPercentages:")
print(df['collection_risk'].value_counts(normalize=True) * 100)

# Visualize
plt.figure(figsize=(10, 6))
colors = ['green', 'orange', 'red']
df['collection_risk'].value_counts().plot(kind='bar', color=colors)
plt.title('Collection Risk Priority Distribution')
plt.xlabel('Collection Priority')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

## 3. EMI and Income Analysis

```python
# Analyze EMI to income ratio
df['emi_income_ratio'] = df['emi_amount'] / df['monthly_income']

print("EMI to Income Ratio Statistics:")
print(df['emi_income_ratio'].describe())

# Analyze by collection risk
print("\nAverage EMI-Income Ratio by Collection Risk:")
print(df.groupby('collection_risk')['emi_income_ratio'].agg(['count', 'mean', 'std']))

# Plot relationship
plt.figure(figsize=(10, 6))
for risk_level in df['collection_risk'].unique():
    data = df[df['collection_risk'] == risk_level]['emi_income_ratio']
    plt.hist(data, alpha=0.6, label=risk_level, bins=20)
plt.xlabel('EMI/Income Ratio')
plt.ylabel('Frequency')
plt.title('EMI-Income Ratio Distribution by Collection Risk')
plt.legend()
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

## 4. Late Payment Analysis

```python
# Analyze late payments
print("Late Payment Statistics by Collection Risk:")
print(df.groupby('collection_risk')[['past_due_days', 'missed_payments']].agg(['count', 'mean', 'max']))

# Visualize
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

ax1 = axes[0]
df.boxplot(column='past_due_days', by='collection_risk', ax=ax1)
ax1.set_title('Past Due Days by Collection Risk')
ax1.set_xlabel('Collection Priority')
ax1.set_ylabel('Past Due Days')

ax2 = axes[1]
df.boxplot(column='missed_payments', by='collection_risk', ax=ax2)
ax2.set_title('Missed Payments by Collection Risk')
ax2.set_xlabel('Collection Priority')
ax2.set_ylabel('Missed Payments')

plt.tight_layout()
plt.show()
```
